## PyTorch - Models

#### Starting with `torch.nn`

`torch.nn` is the neural network module. It contains classes and functions for building and testing deep learning models.
- https://docs.pytorch.org/docs/stable/nn.html

These are some of the basic building blocks for graphs: `Containers`, `Convolution Layers`, `Linear Layers`, `Pooling Layers`, `Transformer Layers`, `Normalization Layers`, `Loss Functions` and more (refer to the link).

At the heart of it is `nn.Module` – base class for all neural network models.

#### Difference between `nn.Parameter` and `nn.Buffer`

These are both used to register tensors in the model, but for different purposes.

- `torch.nn.Parameter` is a subclass of `torch.Tensor` that is automatically registered as a trainable parameter when assigned to a model attribute. This means that it will appear in `model.parameters()` and it will be updated by the optimizer.
- `torch.nn.Buffer` is a kind of Tensor that should not be considered as a model parameter.

```python
import torch
import torch.nn as nn

class MyModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.weight = nn.Parameter(torch.randn(3, 3)) # Learnable 3x3 weight
        self.register_buffer("Constant", torch.ones(3)) # Non-trainable buffer
    
    def forward(self, X):
        return torch.matmul(X, self.weight) * self.register_buffer

model = MyModel()
print(list(model.parameters()))
print(model.Constant)
print(f"State Dict: {model.state_dict().keys()}")
```

#### The `torch.nn` Module Overview

##### 1. `Containers`

Used to build modular models by grouping layers or submodules.
- `nn.Sequential` – Stack layers in order, pass output of one to the next
- `nn.ModuleList` – List of layers, you manually define flow
- `nn.ModuleDict` – Dictionary of layers (useful for dynamic architectures)

##### 2. `Convolution Layers`

Apply convolutional operations—essential for image and signal processing.
- `nn.Conv1d` – 1D data (e.g., time series)
- `nn.Conv2d` – 2D data (e.g., images)
- `nn.Conv3d` – 3D data (e.g., volumetric/MRI)

Variants:
- `nn.ConvTranspose*` – learnable upsampling (used in decoders)



##### 3. `Pooling Layers`

Downsample feature maps to reduce spatial size and computational load.
- `nn.MaxPool1d/2d/3d` – Takes max value
- `nn.AvgPool1d/2d/3d` – Takes average
- `nn.Adaptive*Pool` – Output is fixed size, input can vary

##### 4. `Padding Layers`

Add extra values (e.g., zeros) around data borders.
- `nn.ConstantPad1d/2d/3d`
- `nn.ReflectionPad*`, `nn.ReplicationPad*`

##### 5. `Non-linear Activations` (as Layers)

Used after linear/convolution layers to introduce non-linearity.
- `nn.ReLU`, `nn.Sigmoid`, `nn.Tanh`, `nn.LeakyReLU`, `nn.GELU`, `nn.ELU`
- `nn.Softmax` – convert scores to probabilities

##### 6. `Non-linear Activations` (Functional/Other)

These are sometimes available as functions (in `torch.nn.functional`) or less commonly used.
- `nn.Softplus`, `nn.Softsign`, `nn.SELU`, `nn.Hardtanh`

##### 7. `Normalization Layers`

Normalize activations to improve convergence during training.
- `nn.BatchNorm1d/2d/3d` – normalizes across batch
- `nn.LayerNorm` – normalizes across features
- `nn.GroupNorm`, `nn.InstanceNorm*`

##### 8. `Recurrent Layers`

Used for sequential data like text, speech, or time series.
- `nn.RNN` – vanilla RNN
- `nn.LSTM` – Long Short-Term Memory
- `nn.GRU` – Gated Recurrent Unit

##### 9. `Transformer Layers`

Modern attention-based architecture for sequences.
- `nn.Transformer` – complete transformer model
- `nn.TransformerEncoder/Decoder`, `nn.MultiheadAttention`



##### 10. `Linear Layers`

Basic dense layers, used everywhere.
- `nn.Linear` – fully connected layer: input → output

##### 11. `Dropout Layers`

Randomly zero out elements to prevent overfitting.
- `nn.Dropout`, `nn.Dropout2d/3d`, `nn.AlphaDropout`

##### 12. `Sparse Layers`

Used when dealing with sparse tensors.
- `nn.Embedding` – maps indices to dense vectors
- `nn.EmbeddingBag` – like embedding but handles variable-length sequences efficiently



##### 13. `Distance Functions`

Used in metric learning or custom loss functions.
- `nn.PairwiseDistance`, `nn.CosineSimilarity`

##### 14. `Loss Functions`

Compute a loss value to guide training.
- `nn.CrossEntropyLoss` – classification
- `nn.MSELoss` – regression
- `nn.BCELoss`, `nn.NLLLoss`, `nn.SmoothL1Loss`, etc.



##### 15. `Vision Layers`

Helpers for computer vision models.
- `nn.PixelShuffle` – upsampling layer
- `nn.Upsample` – resize images/tensors
- `nn.Conv2d`, `nn.BatchNorm2d` – most commonly used in vision

##### 16. `Shuffle Layers`

Reorder data spatially.
- `nn.PixelShuffle`, `nn.PixelUnshuffle` – used in super-resolution

##### 17. `DataParallel / Multi-GPU Layers`

Parallelizes model execution across multiple GPUs.
- `nn.DataParallel` – legacy, easy to use but has limitations

Preferred modern approach: use `torch.nn.parallel.DistributedDataParallel` (outside `nn`)

##### 18. `Utilities`

Miscellaneous helpers.
- `nn.Identity` – useful as a placeholder
- `nn.Flatten`, `nn.Unflatten` – reshape tensors
- `nn.Parameter`, `register_buffer()` – custom parameter/buffer management



##### 19. `Quantized Functions`

Used for quantization-aware training (QAT) and post-training quantization.
- `nn.quantized.*` – used when deploying models on edge devices

##### 20. `Lazy Modules`

Used when you don’t know input size until the first forward pass.
- `nn.LazyLinear`, `nn.LazyConv2d`, etc.
